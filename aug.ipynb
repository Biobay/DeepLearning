```python
import matplotlib.pyplot as plt
import numpy as np
from PIL import Image
import os
import pandas as pd
from torchvision import transforms as T

# Assicurati che 'config' sia importato o che i suoi valori siano disponibili
# from src import config

def show_augmentation(image_path, transform, num_examples=5):
    """
    Mostra un'immagine originale e diverse sue versioni aumentate.
    """
    # Carica l'immagine originale
    original_image = Image.open(image_path).convert("RGBA")
    background = Image.new('RGBA', original_image.size, (255, 255, 255))
    original_image = Image.alpha_composite(background, original_image).convert('RGB')

    # Crea una griglia di immagini
    plt.figure(figsize=(15, 5))
    
    # Mostra l'immagine originale
    plt.subplot(1, num_examples + 1, 1)
    plt.imshow(original_image)
    plt.title("Originale")
    plt.axis('off')
    
    # Mostra le versioni aumentate
    for i in range(num_examples):
        augmented_image = transform(original_image)
        # De-normalizza per la visualizzazione
        augmented_image = augmented_image.permute(1, 2, 0) * 0.5 + 0.5
        augmented_image = np.clip(augmented_image, 0, 1)
        
        plt.subplot(1, num_examples + 1, i + 2)
        plt.imshow(augmented_image)
        plt.title(f"Augmented {i+1}")
        plt.axis('off')
        
    plt.suptitle("Esempio di Data Augmentation", fontsize=16, fontweight='bold')
    plt.tight_layout()
    plt.show()

# --- ESEGUI LA VISUALIZZAZIONE ---
# 1. Definisci la stessa pipeline di transform del training
train_transform_viz = T.Compose([
    T.Resize((config.STAGE1_IMAGE_SIZE, config.STAGE1_IMAGE_SIZE)),
    T.RandomHorizontalFlip(p=0.5),
    T.ColorJitter(brightness=0.3, contrast=0.3, saturation=0.3, hue=0.1),
    T.ToTensor(),
    T.Normalize(mean=[0.5, 0.5, 0.5], std=[0.5, 0.5, 0.5])
])

# 2. Scegli un'immagine di esempio
example_image_path = os.path.join(config.IMAGE_DIR, "001.png") # Es. Bulbasaur

# 3. Lancia la funzione
show_augmentation(example_image_path, train_transform_viz)
```
