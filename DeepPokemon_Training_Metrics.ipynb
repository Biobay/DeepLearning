## 4. Avvia il training del modello
Esegui la funzione di training e salva le metriche ottenute durante l'addestramento.

```python
# Esegui il training e salva le metriche
train_losses = []
val_losses = []

# Modifica la funzione train per restituire le metriche oppure usa callback/logging
# Qui si assume che la funzione train aggiorni le liste train_losses e val_losses
train_losses, val_losses = train(config)
```

## 5. Stampa le metriche di training
Visualizza le metriche raccolte durante l'addestramento.

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10,5))
plt.plot(train_losses, label='Training Loss')
plt.xlabel('Epoch')
plt.ylabel('Loss')
plt.title('Andamento della Loss durante il Training')
plt.legend()
plt.show()
```

## 6. Stampa le metriche di validazione
Visualizza le metriche raccolte durante la validazione finale del modello.

```python
plt.figure(figsize=(10,5))
plt.plot(val_losses, label='Validation Loss', color='orange')
plt.xlabel('Epoch')
plt.ylabel('Loss')
plt.title('Andamento della Loss durante la Validazione')
plt.legend()
plt.show()
```
